# Normalization Data

```python
scaler=StandardScaler()
scaler.fit(df)
```

```python
scaler_train=scaler.transform(df)
```

# DBSCAN Clustering

```python
from sklearn.metrics import v_measure_score
from sklearn.cluster import DBSCAN
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN

# Run DBSCAN
dbscan_cluster = DBSCAN(eps=0.5, min_samples=15)
dbscan_cluster.fit(scaler_train)

labels = dbscan_cluster.labels_
unique_labels = set(labels)

# Plotting
plt.figure(figsize=(8,6))

# Assign colors for clusters
colors = plt.cm.tab10(np.linspace(0, 1, len(unique_labels)))

for k, col in zip(unique_labels, colors):
    if k == -1:
        # Noise points (label = -1) → black color
        col = 'k'
        label_name = 'Noise'
    else:
        label_name = f'Cluster {k}'

    # Select the points for this cluster
    class_member_mask = (labels == k)
    xy = scaler_train[class_member_mask]

    plt.scatter(xy[:, 0], xy[:, 1], c=[col], label=label_name, s=30, edgecolors='k')

plt.xlabel("$x_1$")
plt.ylabel("$x_2$")
plt.title("DBSCAN Clustering")
plt.legend()
plt.show()

# Number of clusters and noise
n_clusters = len(unique_labels) - (1 if -1 in labels else 0)
n_noise = list(labels).count(-1)

print(f"Estimated no. of clusters: {n_clusters}")
print(f"Estimated no. of noise points: {n_noise}")
```

```python
from sklearn.metrics import silhouette_score
print('score: ',silhouette_score(scaler_train,dbscan_cluster.labels_))
```
